```python
# --------------------------------------
# ABC Analysis - Product Segmentation
# --------------------------------------

# Sort products by revenue
abc_analysis = product_performance.sort_values(
    "revenue",
    ascending=False
).copy()

# Calculate revenue percentage
total_product_revenue = abc_analysis["revenue"].sum()

abc_analysis["revenue_percentage"] = (
    abc_analysis["revenue"] / total_product_revenue
) * 100

# Calculate cumulative revenue percentage
abc_analysis["cumulative_revenue_percentage"] = (
    abc_analysis["revenue_percentage"].cumsum()
)

# Assign ABC classification
def classify_product(cumulative_percentage):
    if cumulative_percentage <= 80:
        return "A"
    elif cumulative_percentage <= 95:
        return "B"
    else:
        return "C"

abc_analysis["abc_class"] = (
    abc_analysis["cumulative_revenue_percentage"]
    .apply(classify_product)
)

print("ABC Analysis:")
print(
    abc_analysis[
        [
            "item_id",
            "revenue",
            "revenue_percentage",
            "cumulative_revenue_percentage",
            "abc_class"
        ]
    ].head(20)
)
```

```python
# --------------------------------------
# ABC Classification Summary
# --------------------------------------

abc_summary = (
    abc_analysis
    .groupby("abc_class")
    .agg(
        products=("item_id", "count"),
        revenue=("revenue", "sum")
    )
    .reset_index()
)

abc_summary["revenue_percentage"] = (
    abc_summary["revenue"] /
    abc_summary["revenue"].sum()
) * 100

print("ABC Classification Summary:")
print(abc_summary)
```

```python
# --------------------------------------
# Inventory Risk Analysis - Demand Metrics
# --------------------------------------

# Aggregate demand at product level
inventory_analysis = (
    sales_365_long
    .groupby("item_id")
    .agg(
        total_units=("units_sold", "sum"),
        avg_daily_demand=("units_sold", "mean"),
        demand_std=("units_sold", "std"),
        max_daily_demand=("units_sold", "max"),
        revenue=("revenue", "sum")
    )
    .reset_index()
)

# Calculate coefficient of variation
inventory_analysis["coefficient_of_variation"] = (
    inventory_analysis["demand_std"] /
    inventory_analysis["avg_daily_demand"]
)

# Add ABC classification
inventory_analysis = inventory_analysis.merge(
    abc_analysis[["item_id", "abc_class"]],
    on="item_id",
    how="left"
)

print("Inventory Analysis:")
print(inventory_analysis.head(20))
```

```python
# --------------------------------------
# High Demand Variability Products
# --------------------------------------

high_variability_products = (
    inventory_analysis
    .sort_values("coefficient_of_variation", ascending=False)
    .head(20)
)

print("Top 20 High-Variability Products:")
print(high_variability_products)
```

```python
# --------------------------------------
# Inventory Risk Score
# --------------------------------------

# Create numerical weights for ABC classes
abc_weights = {
    "A": 3,
    "B": 2,
    "C": 1
}

inventory_analysis["abc_weight"] = (
    inventory_analysis["abc_class"]
    .map(abc_weights)
)

# Normalize demand and variability
inventory_analysis["demand_score"] = (
    inventory_analysis["avg_daily_demand"] /
    inventory_analysis["avg_daily_demand"].max()
)

inventory_analysis["variability_score"] = (
    inventory_analysis["coefficient_of_variation"] /
    inventory_analysis["coefficient_of_variation"].max()
)

# Calculate inventory risk score
inventory_analysis["inventory_risk_score"] = (
    inventory_analysis["abc_weight"] * 0.40 +
    inventory_analysis["demand_score"] * 0.30 +
    inventory_analysis["variability_score"] * 0.30
)

# Rank products by inventory risk
inventory_risk = (
    inventory_analysis
    .sort_values(
        "inventory_risk_score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("Top 20 Inventory Risk Products:")
print(
    inventory_risk[
        [
            "item_id",
            "abc_class",
            "avg_daily_demand",
            "demand_std",
            "coefficient_of_variation",
            "revenue",
            "inventory_risk_score"
        ]
    ].head(20)
)
```

```python
# --------------------------------------
# Inventory Risk Categories
# --------------------------------------

# Define risk thresholds using the distribution
high_threshold = inventory_risk["inventory_risk_score"].quantile(0.75)
medium_threshold = inventory_risk["inventory_risk_score"].quantile(0.40)

# Assign risk categories
def classify_risk(score):
    if score >= high_threshold:
        return "High Risk"
    elif score >= medium_threshold:
        return "Medium Risk"
    else:
        return "Low Risk"

inventory_risk["risk_category"] = (
    inventory_risk["inventory_risk_score"]
    .apply(classify_risk)
)

# Display top risk products
top_risk_products = (
    inventory_risk[
        [
            "item_id",
            "abc_class",
            "avg_daily_demand",
            "demand_std",
            "coefficient_of_variation",
            "revenue",
            "inventory_risk_score",
            "risk_category"
        ]
    ]
    .head(20)
)

print("Top Inventory Risk Products:")
print(top_risk_products)
```

```python
# --------------------------------------
# Demand Forecasting - Dataset Preparation
# --------------------------------------

# Aggregate sales across stores for each product and date
forecast_data = (
    sales_365_long
    .groupby(["item_id", "date"])
    .agg(
        units_sold=("units_sold", "sum"),
        revenue=("revenue", "sum")
    )
    .reset_index()
)

# Sort by product and date
forecast_data = forecast_data.sort_values(
    ["item_id", "date"]
).reset_index(drop=True)

print("Forecasting Dataset Shape:")
print(forecast_data.shape)

print("\nForecasting Dataset:")
print(forecast_data.head(20))
```

```python
# --------------------------------------
# Demand Forecasting - Feature Engineering
# --------------------------------------

# Create calendar features
forecast_data["day_of_week"] = forecast_data["date"].dt.dayofweek
forecast_data["month"] = forecast_data["date"].dt.month
forecast_data["week_of_year"] = forecast_data["date"].dt.isocalendar().week.astype(int)

# Create lag features
forecast_data["lag_1"] = (
    forecast_data
    .groupby("item_id")["units_sold"]
    .shift(1)
)

forecast_data["lag_7"] = (
    forecast_data
    .groupby("item_id")["units_sold"]
    .shift(7)
)

forecast_data["lag_28"] = (
    forecast_data
    .groupby("item_id")["units_sold"]
    .shift(28)
)

# Create rolling demand features
forecast_data["rolling_7"] = (
    forecast_data
    .groupby("item_id")["units_sold"]
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)

forecast_data["rolling_28"] = (
    forecast_data
    .groupby("item_id")["units_sold"]
    .transform(
        lambda x: x.shift(1).rolling(28).mean()
    )
)

print("Forecasting Features Created:")
print(forecast_data.head(20))
```

```python
# --------------------------------------
# Remove Rows with Missing Lag Values
# --------------------------------------

forecast_model_data = forecast_data.dropna().copy()

print("Original Dataset Shape:")
print(forecast_data.shape)

print("\nModel Dataset Shape:")
print(forecast_model_data.shape)

print("\nMissing Values:")
print(
    forecast_model_data[
        [
            "lag_1",
            "lag_7",
            "lag_28",
            "rolling_7",
            "rolling_28"
        ]
    ].isnull().sum()
)
```

```python
# --------------------------------------
# Demand Forecasting - Train/Test Split
# --------------------------------------

# Sort data by date
forecast_model_data = forecast_model_data.sort_values(
    "date"
).reset_index(drop=True)

# Use the last 30 days as the test period
test_days = 30

# Find the cutoff date
cutoff_date = (
    forecast_model_data["date"].max()
    - pd.Timedelta(days=test_days - 1)
)

# Create training and test datasets
train_data = forecast_model_data[
    forecast_model_data["date"] < cutoff_date
].copy()

test_data = forecast_model_data[
    forecast_model_data["date"] >= cutoff_date
].copy()

print("Training Period:")
print(train_data["date"].min(), "to", train_data["date"].max())

print("\nTesting Period:")
print(test_data["date"].min(), "to", test_data["date"].max())

print("\nTraining Shape:")
print(train_data.shape)

print("\nTesting Shape:")
print(test_data.shape)
```

```python
# --------------------------------------
# Demand Forecasting - Features and Target
# --------------------------------------

# Features used by the forecasting model
features = [
    "day_of_week",
    "month",
    "week_of_year",
    "lag_1",
    "lag_7",
    "lag_28",
    "rolling_7",
    "rolling_28"
]

# Target variable
target = "units_sold"

# Create training features and target
X_train = train_data[features]
y_train = train_data[target]

# Create testing features and target
X_test = test_data[features]
y_test = test_data[target]

print("Training Features Shape:")
print(X_train.shape)

print("\nTraining Target Shape:")
print(y_train.shape)

print("\nTesting Features Shape:")
print(X_test.shape)

print("\nTesting Target Shape:")
print(y_test.shape)

print("\nFeatures Used:")
print(features)
```

```python
# --------------------------------------
# Demand Forecasting - Naive Baseline
# --------------------------------------

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Use yesterday's demand as the prediction
naive_predictions = X_test["lag_1"]

# Calculate evaluation metrics
naive_mae = mean_absolute_error(
    y_test,
    naive_predictions
)

naive_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        naive_predictions
    )
)

print("Naive Forecast Performance:")
print(f"MAE:  {naive_mae:.4f}")
print(f"RMSE: {naive_rmse:.4f}")
```

```python
# --------------------------------------
# Demand Forecasting - Random Forest Regressor
# --------------------------------------

from sklearn.ensemble import RandomForestRegressor

# Create Random Forest Regressor
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(
    X_train,
    y_train
)

# Generate predictions on test data
rf_predictions = rf_model.predict(X_test)

print("Random Forest Regressor training completed.")
```

```python
# --------------------------------------
# Random Forest - Model Evaluation
# --------------------------------------

# Calculate MAE
rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

# Calculate RMSE
rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

print("Random Forest Regressor Performance:")
print(f"MAE:  {rf_mae:.4f}")
print(f"RMSE: {rf_rmse:.4f}")

print("\nComparison with Naive Forecast:")
print(f"Naive MAE:  {naive_mae:.4f}")
print(f"Naive RMSE: {naive_rmse:.4f}")
```

Random Forest substantially outperforms the Naive Forecast baseline for RetailIQ's 30-day demand forecasting test period.

RetailIQ combines ABC classification, demand intensity, and demand variability to identify products that require different inventory actions.

### Inventory Risk Factors

- **ABC Class:** Represents the business importance of each product based on revenue contribution.
- **Average Daily Demand:** Measures the typical number of units sold per day.
- **Demand Standard Deviation:** Measures how much daily demand fluctuates.
- **Coefficient of Variation:** Measures demand variability relative to average demand.
- **Safety Stock:** Estimates additional inventory needed to absorb demand variability.
- **Reorder Point:** Estimates the inventory level at which replenishment should be considered.

### Stockout Risk Score

The stockout risk score combines:

- **40% ABC business importance**
- **35% demand intensity**
- **25% demand variability**

This prioritizes products that are both commercially important and exposed to demand pressure.

### Inventory Actions

| Risk Score  | Recommended Action       |
|-------------|--------------------------|
| ≥ 0.50      | Prioritize Replenishment |
| 0.25 – 0.49 | Monitor Inventory        |
| < 0.25      | Standard Replenishment   |

### Key Business Insight

RetailIQ identifies a focused group of high-priority products rather than treating every product equally. This allows inventory teams to concentrate replenishment attention on products with the greatest combination of business importance, demand, and variability.

```python
# --------------------------------------
# RetailIQ - Executive KPI Summary
# --------------------------------------

executive_kpis = con.execute("""
    WITH business AS (
        SELECT
            COUNT(DISTINCT item_id) AS active_products,
            COUNT(DISTINCT store_id) AS active_stores,
            COUNT(DISTINCT date) AS active_sales_days,
            SUM(units_sold) AS total_units_sold,
            SUM(revenue) AS total_revenue
        FROM fact_sales
    ),

    forecast AS (
        SELECT
            ROUND(
                AVG(ABS(actual_units - predicted_units)),
                4
            ) AS forecast_mae,

            ROUND(
                SQRT(
                    AVG(
                        POWER(
                            actual_units - predicted_units,
                            2
                        )
                    )
                ),
                4
            ) AS forecast_rmse
        FROM fact_forecast
    ),

    risk AS (
        SELECT
            COUNT(*) FILTER (
                WHERE risk_category = 'High Risk'
            ) AS high_risk_products
        FROM fact_inventory
    )

    SELECT
        business.active_products,
        business.active_stores,
        business.active_sales_days,
        business.total_units_sold,
        ROUND(business.total_revenue, 2) AS total_revenue,
        forecast.forecast_mae,
        forecast.forecast_rmse,
        risk.high_risk_products
    FROM business
    CROSS JOIN forecast
    CROSS JOIN risk
""").df()

display(executive_kpis)
```

```python
# --------------------------------------
# RetailIQ - Underperforming Stores
# --------------------------------------

underperforming_stores = con.execute("""
    WITH store_revenue AS (
        SELECT
            store_id,
            SUM(revenue) AS total_revenue
        FROM fact_sales
        GROUP BY store_id
    ),

    average_store AS (
        SELECT
            AVG(total_revenue) AS avg_store_revenue
        FROM store_revenue
    )

    SELECT
        s.store_id,
        ROUND(s.total_revenue, 2) AS total_revenue,
        ROUND(a.avg_store_revenue, 2) AS average_store_revenue,
        ROUND(
            s.total_revenue - a.avg_store_revenue,
            2
        ) AS revenue_gap,
        ROUND(
            100.0 *
            (s.total_revenue - a.avg_store_revenue)
            / NULLIF(a.avg_store_revenue, 0),
            2
        ) AS gap_pct
    FROM store_revenue s
    CROSS JOIN average_store a
    WHERE s.total_revenue < a.avg_store_revenue
    ORDER BY revenue_gap
""").df()

display(underperforming_stores)
```

```python
# --------------------------------------
# RetailIQ - Forecast Underestimation
# --------------------------------------

forecast_underestimation = con.execute("""
    SELECT
        item_id,
        SUM(actual_units) AS actual_30d_units,
        ROUND(SUM(predicted_units), 2) AS predicted_30d_units,
        ROUND(
            SUM(predicted_units) - SUM(actual_units),
            2
        ) AS forecast_gap,
        ROUND(
            100.0 *
            (SUM(predicted_units) - SUM(actual_units))
            / NULLIF(SUM(actual_units), 0),
            2
        ) AS forecast_gap_pct
    FROM fact_forecast
    GROUP BY item_id
    HAVING SUM(actual_units) > 0
       AND SUM(predicted_units) < SUM(actual_units)
    ORDER BY forecast_gap ASC
    LIMIT 20
""").df()

display(forecast_underestimation)
```

```python
# --------------------------------------
# RetailIQ - Forecast Overestimation
# --------------------------------------

forecast_overestimation = con.execute("""
    SELECT
        item_id,
        SUM(actual_units) AS actual_30d_units,
        ROUND(SUM(predicted_units), 2) AS predicted_30d_units,
        ROUND(
            SUM(predicted_units) - SUM(actual_units),
            2
        ) AS forecast_gap,
        ROUND(
            100.0 *
            (SUM(predicted_units) - SUM(actual_units))
            / NULLIF(SUM(actual_units), 0),
            2
        ) AS forecast_gap_pct
    FROM fact_forecast
    GROUP BY item_id
    HAVING SUM(actual_units) > 0
       AND SUM(predicted_units) > SUM(actual_units)
    ORDER BY forecast_gap DESC
    LIMIT 20
""").df()

display(forecast_overestimation)
```

```python
# --------------------------------------
# RetailIQ - Store Revenue Opportunity
# --------------------------------------

store_opportunity = con.execute("""
    WITH store_revenue AS (
        SELECT
            store_id,
            SUM(revenue) AS total_revenue
        FROM fact_sales
        GROUP BY store_id
    ),

    average_store AS (
        SELECT
            AVG(total_revenue) AS avg_revenue
        FROM store_revenue
    )

    SELECT
        s.store_id,
        ROUND(s.total_revenue, 2) AS total_revenue,
        ROUND(a.avg_revenue, 2) AS average_revenue,
        ROUND(
            a.avg_revenue - s.total_revenue,
            2
        ) AS opportunity_gap,
        CASE
            WHEN s.total_revenue < a.avg_revenue
                THEN 'Growth Opportunity'
            ELSE 'At or Above Average'
        END AS opportunity_type
    FROM store_revenue s
    CROSS JOIN average_store a
    ORDER BY opportunity_gap DESC
""").df()

display(store_opportunity)
```

```python
# --------------------------------------
# RetailIQ - Store Category Performance
# --------------------------------------

store_category_analysis = con.execute("""
    SELECT
        f.store_id,
        p.cat_id,
        SUM(f.units_sold) AS total_units_sold,
        ROUND(SUM(f.revenue), 2) AS total_revenue,
        ROUND(
            100.0 * SUM(f.revenue)
            / NULLIF(
                SUM(SUM(f.revenue))
                OVER (PARTITION BY f.store_id),
                0
            ),
            2
        ) AS store_category_share_pct
    FROM fact_sales f
    JOIN dim_product p
        ON f.item_id = p.item_id
    GROUP BY
        f.store_id,
        p.cat_id
    ORDER BY
        f.store_id,
        total_revenue DESC
""").df()

display(store_category_analysis)
```

```python
# --------------------------------------
# RetailIQ - Product-Level Store Opportunity
# --------------------------------------

product_store_opportunity = con.execute("""
    WITH store_product_revenue AS (
        SELECT
            item_id,
            store_id,
            SUM(units_sold) AS total_units_sold,
            SUM(revenue) AS total_revenue
        FROM fact_sales
        GROUP BY
            item_id,
            store_id
    ),

    comparison AS (
        SELECT
            ca3.item_id,

            ca3.total_units_sold AS units_CA3,
            ca4.total_units_sold AS units_CA4,

            ca3.total_revenue AS revenue_CA3,
            ca4.total_revenue AS revenue_CA4

        FROM store_product_revenue ca3

        JOIN store_product_revenue ca4
            ON ca3.item_id = ca4.item_id

        WHERE ca3.store_id = 'CA_3'
          AND ca4.store_id = 'CA_4'
    )

    SELECT
        item_id,
        units_CA3,
        units_CA4,
        ROUND(revenue_CA3, 2) AS revenue_CA3,
        ROUND(revenue_CA4, 2) AS revenue_CA4,
        ROUND(
            revenue_CA3 - revenue_CA4,
            2
        ) AS revenue_gap_CA3_vs_CA4,
        units_CA3 - units_CA4 AS units_gap
    FROM comparison
    ORDER BY revenue_gap_CA3_vs_CA4 DESC
    LIMIT 20
""").df()

display(product_store_opportunity)
```

```python
# --------------------------------------
# RetailIQ - Final Product Opportunity
# --------------------------------------

final_product_opportunity = con.execute("""
    WITH store_product_revenue AS (
        SELECT
            item_id,
            store_id,
            SUM(units_sold) AS total_units_sold,
            SUM(revenue) AS total_revenue
        FROM fact_sales
        GROUP BY
            item_id,
            store_id
    ),

    store_comparison AS (
        SELECT
            ca3.item_id,

            ca3.total_units_sold AS units_CA3,
            ca4.total_units_sold AS units_CA4,

            ca3.total_revenue AS revenue_CA3,
            ca4.total_revenue AS revenue_CA4

        FROM store_product_revenue ca3

        JOIN store_product_revenue ca4
            ON ca3.item_id = ca4.item_id

        WHERE ca3.store_id = 'CA_3'
          AND ca4.store_id = 'CA_4'
    )

    SELECT
        c.item_id,

        i.abc_class,

        ROUND(c.revenue_CA3, 2) AS revenue_CA3,

        ROUND(c.revenue_CA4, 2) AS revenue_CA4,

        ROUND(
            c.revenue_CA3 - c.revenue_CA4,
            2
        ) AS revenue_gap,

        ROUND(i.avg_daily_demand, 2)
            AS avg_daily_demand,

        ROUND(i.inventory_risk_score, 3)
            AS inventory_risk_score,

        i.risk_category

    FROM store_comparison c

    JOIN fact_inventory i
        ON c.item_id = i.item_id

    WHERE c.revenue_CA3 > c.revenue_CA4

    ORDER BY
        revenue_gap DESC

    LIMIT 20
""").df()

display(final_product_opportunity)
```

```python
# --------------------------------------
# RetailIQ - Final Recommendation Dataset
# --------------------------------------

final_recommendations = con.execute("""
    WITH store_product_revenue AS (
        SELECT
            item_id,
            store_id,
            SUM(revenue) AS total_revenue
        FROM fact_sales
        GROUP BY
            item_id,
            store_id
    ),

    product_comparison AS (
        SELECT
            ca3.item_id,

            ca3.total_revenue AS revenue_CA3,
            ca4.total_revenue AS revenue_CA4,

            ca3.total_revenue - ca4.total_revenue
                AS revenue_gap

        FROM store_product_revenue ca3

        JOIN store_product_revenue ca4
            ON ca3.item_id = ca4.item_id

        WHERE ca3.store_id = 'CA_3'
          AND ca4.store_id = 'CA_4'
    ),

    forecast_summary AS (
        SELECT
            item_id,

            SUM(actual_units) AS actual_30d_units,

            SUM(predicted_units)
                AS predicted_30d_units,

            SUM(predicted_units)
                - SUM(actual_units)
                AS forecast_gap

        FROM fact_forecast

        GROUP BY item_id
    )

    SELECT

        p.item_id,

        i.abc_class,

        ROUND(p.revenue_CA3, 2)
            AS revenue_CA3,

        ROUND(p.revenue_CA4, 2)
            AS revenue_CA4,

        ROUND(p.revenue_gap, 2)
            AS revenue_gap,

        ROUND(i.avg_daily_demand, 2)
            AS avg_daily_demand,

        ROUND(i.inventory_risk_score, 3)
            AS inventory_risk_score,

        i.risk_category,

        f.actual_30d_units,

        ROUND(f.predicted_30d_units, 2)
            AS predicted_30d_units,

        ROUND(f.forecast_gap, 2)
            AS forecast_gap,

        CASE

            WHEN
                i.abc_class = 'A'
                AND i.risk_category = 'High Risk'
                AND p.revenue_gap > 30000
            THEN 'Critical Opportunity'

            WHEN
                i.abc_class = 'A'
                AND i.risk_category = 'High Risk'
                AND p.revenue_gap > 15000
            THEN 'High Priority'

            WHEN
                i.risk_category = 'High Risk'
            THEN 'Inventory Attention'

            WHEN
                p.revenue_gap > 15000
            THEN 'Store Growth Opportunity'

            ELSE 'Monitor'

        END AS recommendation

    FROM product_comparison p

    JOIN fact_inventory i
        ON p.item_id = i.item_id

    LEFT JOIN forecast_summary f
        ON p.item_id = f.item_id

    ORDER BY
        p.revenue_gap DESC

    LIMIT 50
""").df()

display(final_recommendations)
```
